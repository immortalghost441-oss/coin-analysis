```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('crypto_market.csv')
df
```

```python

df = df.sort_values('date')
n=df.groupby('date')['price_usd'].mean()
date=[]
date=df['date']
price_usd=df['price_usd']
```

```python
plt.bar(date,price_usd)
plt.xticks(rotation=45)
plt.xlabel("date")
plt.ylabel('price')
plt.title('coin price')
```

```python

plt.bar(df["coin"], df["price_usd"])
plt.xticks(rotation=45)
plt.ylabel("price_usd")
plt.xlabel('coin')
plt.title('coin price')
```

```python
plt.bar(df["coin"], df["market_cap"])
plt.ylabel("market cap")
plt.xlabel('coin')
plt.title('coin price')
```

```python
import numpy as np
plt.scatter(df["market_cap"], df["price_usd"],alpha=0.7)

plt.xscale("log")
plt.yscale("log")

plt.xlabel("Market Cap (log)")
plt.ylabel("Price (log)")
plt.title("Market Cap vs Price Relationship")
plt.legend()
plt.show()
```

```python
for coin in df["coin"].unique():
    temp = df[df["coin"] == coin]
    plt.scatter(temp["market_cap"], temp["price_usd"], label=coin,alpha=0.7)

plt.xscale("log")
plt.yscale("log")

plt.xlabel("Market Cap (log)")
plt.ylabel("Price USD (log)")
plt.legend()
plt.show()
```

```python
plt.scatter(df["price_usd"], df["fear_greed_index"])

plt.xscale("log")
plt.yscale("log")
plt.ylabel("fear_greed_index")
plt.xlabel('price_usd')
plt.title('prive vs fear n greed')
```
